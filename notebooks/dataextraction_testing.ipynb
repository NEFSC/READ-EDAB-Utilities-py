```python
# 1. Helper to extract data from the legacy compiled CSV
def extract_legacy_timeseries(filepath, prod, period_code, stat_var='AMEAN', alg=None, subareas=None, file_type='STACKED_STATS'):
    df = pd.read_csv(filepath)
    mask = (df['PROD'] == prod) & (df['PERIOD_CODE'] == period_code) & (df['FILE_TYPE'] == file_type)
    
    if alg: mask = mask & (df['ALG'] == alg)
    if subareas: mask = mask & (df['SUBAREA'].isin(subareas))
        
    df_filtered = df[mask].copy()
    if df_filtered.empty: return pd.DataFrame()
        
    def parse_period(period_str):
        if not isinstance(period_str, str): return pd.NaT
        try:
            # 1. Standard Timeseries parsing
            if period_str.startswith('A_'): 
                return pd.to_datetime(period_str[2:], format='%Y')
            elif period_str.startswith('M_'): 
                return pd.to_datetime(f"{period_str[2:6]}-{period_str[6:8]}-15")
            elif period_str.startswith('W_'): 
                return pd.to_datetime(f"{period_str[2:6]}{period_str[6:8]}-0", format="%Y%W-%w")
            
            # 2. 🎯 FIX: Climatology parsing (assigning to generic year 1997 to match pipeline dates)
            elif period_str.startswith('MONTH_'):
                mm = period_str.split('_')[1]
                return pd.to_datetime(f"1997-{mm}-01") 
            elif period_str.startswith('WEEK_'):
                ww = period_str.split('_')[1]
                return pd.to_datetime(f"1997{ww}-0", format="%Y%W-%w")
            elif period_str.startswith('ANNUAL_'):
                return pd.to_datetime("1997-01-01")
                
        except: return pd.NaT
        return pd.NaT
        
    df_filtered['Date'] = df_filtered['PERIOD'].apply(parse_period)
    cols_to_keep = ['Date', 'SUBAREA', 'PROD', 'ALG', 'PERIOD', stat_var]
    return df_filtered.dropna(subset=['Date'])[cols_to_keep].sort_values(['SUBAREA', 'Date']).reset_index(drop=True)


# 2. Main Comparison Orchestrator
def plot_pipeline_vs_legacy(extracted_files, legacy_file):
    subareas = ['GB', 'GOM', 'MAB']
    
    legacy_prod_map = {'CHL': 'CHLOR_A', 'PPD': 'PPD', 'PP_Eppley': 'PPD'}
    legacy_alg_map = {'CHL': 'CCI', 'PPD': 'VGPM2', 'PP_Eppley': 'VGPM2'}
    legacy_type_map = {'STATS': 'STACKED_STATS', 'ANOMS': 'STACKED_ANOMS', 'ANNUAL': 'STACKED_STATS'}

    for filepath in extracted_files:
        filename = os.path.basename(filepath)
        print(f"\n{'='*70}")
        print(f"📊 Processing File: {filename}")
        
        if 'A-OCCCI-PPD-ANNUAL' in filename:
            print("  ⏩ Skipping A-OCCCI-PPD-ANNUAL file as requested.")
            continue
        
        parts = filename.split('-')
        if len(parts) < 5:
            print("  ⚠ Filename doesn't match expected structure. Skipping.")
            continue
            
        period_str = parts[0]
        data_type = parts[3]
        
        # 🎯 FIX: Exact 1:1 matching for periods! 'MONTH' == 'MONTH', 'M' == 'M'
        leg_period = period_str 
        leg_file_type = legacy_type_map.get(data_type, 'STACKED_STATS')
        
        # Load New Pipeline Data
        header_line = 0
        with open(filepath, 'r') as f:
            for i, line in enumerate(f):
                if line.startswith("Date,period_code"):
                    header_line = i
                    break
                    
        df_new = pd.read_csv(filepath, skiprows=header_line)
        df_new['Date'] = pd.to_datetime(df_new['Date'])
        
        target_stat = '_Sum' if data_type == 'ANNUAL' else '_Mean'
        
        # 🎯 FIX: Safely exclude geometric stats
        target_cols = [
            col for col in df_new.columns 
            if col.endswith(target_stat) 
            and 'Geom' not in col 
            and not col.startswith('PP_VGPM')
        ]
        
        if not target_cols:
            print(f"  ⚠ Skipping - Could not identify any '{target_stat}' columns in the new data.")
            continue
            
        print(f"  -> Found {len(df_new)} Pipeline records with {len(target_cols)} target variables.")

        for new_stat_col in target_cols:
            base_var = new_stat_col.replace(target_stat, '')
            
            if 'PSC_' in base_var:
                leg_prod = base_var.upper()
                leg_alg = None 
            else:
                leg_prod = legacy_prod_map.get(base_var, base_var)
                leg_alg = legacy_alg_map.get(base_var)
                
            print(f"\n  🎯 Variable: {new_stat_col} (Mapped to Legacy PROD: {leg_prod})")
            
            # Fetch Legacy Data
            df_leg = extract_legacy_timeseries(
                legacy_file, prod=leg_prod, period_code=leg_period, 
                stat_var='AMEAN', alg=leg_alg, subareas=subareas, file_type=leg_file_type
            )
            print(f"    -> Found {len(df_leg)} matching Legacy records.")

            # Generate the Plot
            fig, axes = plt.subplots(3, 1, figsize=(12, 12), sharex=True)
            fig.suptitle(f"{base_var} {data_type} ({period_str}) - Pipeline vs Legacy", fontsize=16, y=0.95)

            for i, subarea in enumerate(subareas):
                ax = axes[i]
                new_sub = df_new[df_new['subarea'] == subarea].sort_values('Date')
                
                # 1. Plot New Data
                if not new_sub.empty:
                    ax.plot(new_sub['Date'], new_sub[new_stat_col], marker='o', linestyle='-', linewidth=1.5, color='royalblue', label=f'New Pipeline')
                
                # 2. Plot Legacy Data
                leg_sub = pd.DataFrame()
                if not df_leg.empty:
                    leg_sub = df_leg[df_leg['SUBAREA'] == subarea].sort_values('Date')
                    if not leg_sub.empty:
                        ax.plot(leg_sub['Date'], leg_sub['AMEAN'], marker='s', linestyle='--', linewidth=1.5, color='darkorange', label='Legacy Data')
                
                ax.set_title(f'Subarea: {subarea}')
                ax.grid(True, linestyle=':', alpha=0.7)
                if not new_sub.empty or not leg_sub.empty:
                    ax.legend(loc='upper right')
                
                if data_type == 'ANNUAL':
                    ax.ticklabel_format(style='sci', axis='y', scilimits=(0,0))

            plt.xlabel('Date', fontsize=12)
            plt.tight_layout(rect=[0, 0, 1, 0.93])
            plt.show()

# 3. Execution
legacy_file = '/Users/kimberly.hyde/Documents/nadata/PROJECTS/SOE-Phytoplankton/V2026/DATA_EXTRACTS/V2026-NES_EPU_NOESTUARIES-COMPILED_DATA_FILE.CSV'

# Replace with your actual list of files
extracted_files = get_extracted_files('OCCCI','NES_EPU_NOESTUARIES',data_type='STATS')

plot_pipeline_vs_legacy(extracted_files, legacy_file)
```

```python


# 1. Safely Parse the Legacy CSV
# (Bypasses parsing errors caused by unquoted commas in metadata columns)
def parse_legacy_data(filepath):
    data = []
    with open(filepath, 'r') as f:
        f.readline() # Skip header
        for line in f:
            parts = line.split(',')
            prod = parts[1]
            alg = parts[2]
            
            # Target only PPD and VGPM2
            if prod == 'PPD' and alg == 'VGPM2':
                data.append({
                    'Year': int(parts[4]),
                    'subarea': parts[6],
                    'Legacy_Annual_Sum': float(parts[11]) if parts[11].strip() else np.nan
                })
    return pd.DataFrame(data)

df_legacy = parse_legacy_data(
    '/Users/kimberly.hyde/Documents/nadata/PROJECTS/SOE-Phytoplankton/V2026/PPREQ_EXTRACTS/NES_EPU_NOESTUARIES/FINAL_MERGED_SUMS/MERGED_ANNUAL_SUM-NES_EPU_NOESTUARIES-PPD-VGPM2_CHLOR_A-CCI-STATS-V2026.CSV'
    )

# 1. Define the file path
pipeline_file = '/Users/kimberly.hyde/Documents/nadata/DATASETS/OCCCI/V6.0/EXTRACTED_DATA/NES_EPU_NOESTUARIES/A-OCCCI-PPD-ANNUAL-NES_EPU_NOESTUARIES-EXTRACTED_DATA.csv'

# 2. Dynamically find the real header row
header_line = 0
with open(pipeline_file, 'r') as f:
    for i, line in enumerate(f):
        # The true data header always starts exactly like this
        if line.startswith("Date,period_code"):
            header_line = i
            break

# 3. Load the dataframe using the dynamically discovered row index
df_new = pd.read_csv(pipeline_file, skiprows=header_line)

# 4. Extract the Year safely!
df_new['Year'] = pd.to_datetime(df_new['Date']).dt.year

# 3. Merge the datasets for plotting
subareas = ['GB', 'GOM', 'MAB']
fig, axes = plt.subplots(3, 1, figsize=(12, 15), sharex=True)
fig.suptitle('Annual Total Primary Productivity Comparison (Pipeline vs Legacy)', fontsize=16, y=0.92)

for i, subarea in enumerate(subareas):
    # Filter both dataframes by subarea
    new_sub = df_new[df_new['subarea'] == subarea]
    leg_sub = df_legacy[df_legacy['subarea'] == subarea]
    
    # Merge them on Year so they perfectly align on the X-axis
    merged = pd.merge(new_sub[['Year', 'PP_Eppley_Annual_Total_Sum']], 
                      leg_sub[['Year', 'Legacy_Annual_Sum']], 
                      on='Year', how='outer').sort_values('Year')
    
    # Plotting
    ax = axes[i]
    ax.plot(merged['Year'], merged['PP_Eppley_Annual_Total_Sum'], 
            marker='o', linestyle='-', linewidth=2, color='royalblue', label='New Pipeline (Eppley)')
    
    ax.plot(merged['Year'], merged['Legacy_Annual_Sum'], 
            marker='s', linestyle='--', linewidth=2, color='darkorange', label='Legacy Data (VGPM2)')
    
    ax.set_title(f'Subarea: {subarea}')
    ax.set_ylabel('Total Annual PP (gC yr⁻¹)')
    ax.grid(True, linestyle=':', alpha=0.7)
    ax.legend(loc='upper right')
    
    # Format the Y-axis to scientific notation since carbon sums are massive
    ax.ticklabel_format(style='sci', axis='y', scilimits=(0,0))

plt.xlabel('Year', fontsize=12)
plt.tight_layout(rect=[0, 0, 1, 0.9])
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Safely Parse the Legacy CSV for Area
def parse_legacy_area(filepath):
    data = []
    with open(filepath, 'r') as f:
        f.readline() # Skip header
        for line in f:
            parts = line.split(',')
            prod = parts[1]
            alg = parts[2]
            
            # Target only PPD and VGPM2
            if prod == 'PPD' and alg == 'VGPM2':
                data.append({
                    'Year': int(parts[4]),
                    'subarea': parts[6],
                    # TOTAL_PIXEL_AREA_KM2 is at index 8
                    'Legacy_Area_km2': float(parts[8]) if parts[8].strip() else np.nan
                })
    return pd.DataFrame(data)

df_legacy = parse_legacy_area(
    '/Users/kimberly.hyde/Documents/nadata/PROJECTS/SOE-Phytoplankton/V2026/PPREQ_EXTRACTS/NES_EPU_NOESTUARIES/FINAL_MERGED_SUMS/MERGED_ANNUAL_SUM-NES_EPU_NOESTUARIES-PPD-VGPM2_CHLOR_A-CCI-STATS-V2026.CSV'
    )

# 1. Define the file path
pipeline_file = '/Users/kimberly.hyde/Documents/nadata/DATASETS/OCCCI/V6.0/EXTRACTED_DATA/NES_EPU_NOESTUARIES/A-OCCCI-PPD-ANNUAL-NES_EPU_NOESTUARIES-EXTRACTED_DATA.csv'

# 2. Load the New Pipeline CSV
header_line = 0
with open(pipeline_file, 'r') as f:
    for i, line in enumerate(f):
        if line.startswith("Date,period_code"):
            header_line = i
            break

df_new = pd.read_csv(pipeline_file, skiprows=header_line)
df_new['Year'] = pd.to_datetime(df_new['Date']).dt.year

# 3. Merge and Plot the Area Data
subareas = ['GB', 'GOM', 'MAB']
fig, axes = plt.subplots(3, 1, figsize=(12, 16), sharex=True)
fig.suptitle('Total Extraction Area Comparison (Pipeline vs Legacy)', fontsize=16, y=0.92)

for i, subarea in enumerate(subareas):
    new_sub = df_new[df_new['subarea'] == subarea]
    leg_sub = df_legacy[df_legacy['subarea'] == subarea]
    
    # Merge them on Year
    merged = pd.merge(new_sub[['Year', 'Total_Area_km2']], 
                      leg_sub[['Year', 'Legacy_Area_km2']], 
                      on='Year', how='outer').sort_values('Year')
    
    # Plotting
    ax = axes[i]
    ax.plot(merged['Year'], merged['Total_Area_km2'], 
            marker='o', linestyle='-', linewidth=2, color='royalblue', label='New Pipeline Area')
    
    ax.plot(merged['Year'], merged['Legacy_Area_km2'], 
            marker='s', linestyle='--', linewidth=2, color='darkorange', label='Legacy Area')
    
    ax.set_title(f'Subarea: {subarea}', pad=10)
    ax.set_ylabel('Total Area (km²)')
    ax.grid(True, linestyle=':', alpha=0.7)
    ax.legend(loc='upper right')

plt.xlabel('Year', fontsize=12)
plt.tight_layout(rect=[0, 0, 1, 0.9])
plt.show()
```
